# 클래스 객체
- 클래스에 대해 배워보자.

클래스는 설계도와 같고, 인스턴스는 설계도를 복제한것과 같다.  
클래스는 보통 Camel 형식으로 선언한다. (AaaaaBbbbb)

```python
class Family:				# 클래스 객체 선언 (하드디스크에 저장)
	family = "grandpa"	 	# Attribute(속성) = State(값)
	def print_self(self):	# 메소드
		print(self)			# 속성과 메소드는 '멤버'라고 부름

LEE = Family()              # 인스턴스 선언 (클래스를 메모리에 객체화)
LEE2 = Family()             # 여러 인스턴스 선언 가능
print(Family.family)		# 속성 호출

Family.print_self("Hello")	# 메소드 호출. 인수를 전달하지 않으면 오류 발생
LEE.print_self()	        # self에 인스턴스 객체 LEE를 기본으로 설정함.
LEE2.print_self()	        # LEE와 LEE2는 다른 메모리에 저장됨
```

생성자 함수 (\_\_init\_\_)를 통해 인스턴스 속성값을 정의/초기화할 수 있다.  

```python
class Color:
	red, green, blue = 0, 0, 0  #클래스의 속성값
	thick = 12

	def print_self(self):
		print(self)
	
	def __init__(self, red, green, blue):   #생성자 함수. self는 필수
		self.red = red          # 인스턴스의 속성값 설정
		self.green = green
		self.blue = blue
		self.bright = "max"     # 인스턴스에만 생기는 속성

Purple = Color(128, 0, 128)     #생성자 인수대로 넣어야 선언가능
Purple.thick = 15               # 값 변경
print(f"Red : {Purple.red}, 굵기 : {Purple.thick}, 밝기 : {Purple.bright}")
print(f"클래스 Red : {Color.red}, 굵기 : {Color.thick}, 밝기는 출력 불가")
```

# 상속 (Inheritance)
클래스의 데이터를 또 다른 클래스의 데이터에 넣을 수 있다.  
부모 클래스와, 부모 클래스의 데이터를 상속받은 자식 클래스로 나뉜다.

```python
class RedColor(Color):              # Color 클래스 상속
    def set_GB(self, GB):           # 자식클래스만의 멤버 선언
        self.GB = GB
        self.green = GB
        self.blue = GB
    
    def __init__(self, red, GB = 0):    # 부모클래스 메소드 재정의 (Override)
        self.red = red
        self.set_GB(GB)
        self.bright = "max"

RedColor.print_self("HI")           # 부모클래스의 데이터가 제대로 전달됨
print(RedColor.thick)

Red = RedColor(64, 32)              # Override가 정상 작동됨
print(f"Red를 R{Red.red} G{Red.green} B{Red.blue} (GB{Red.GB})로 설정")
```

# 접근 지정자 (Access Specifiers)
클래스 멤버는 기본적으로 Public으로 설정되어 객체 외부에서도 접근 가능하다.  
멤버 앞에 \_\_ 를 붙여 Private로 설정할 수 있다.  
Private 멤버는 객체 내부에서만 접근 가능하다.

```python
class Student:
    __schoolName = "XYZ School"

    def __init__(self, name, age):
        self.name = name
        self.__salary = age
    def __display(self):
	      print('This is private method.')

std = Student("Bill", 25)
print(std.name)     # 정상 작동
# std.__salary      # 오류 발생.
print(std._Student__salary)   # 직접 접근
```

# Decorators
그 밖에 decorator를 확인하자. 쓰는 것만 썼다.

```python
class Student:
    name = "unknown"
    def __init__(self, name):
        self.age = 20
        self.name = name

    @classmethod            # 인스턴스 객체 대신 클래스 객체를 받는다.
    def tostring(cls):
        print(cls.name)

std = Student("test")
std.tostring()              # std.tostring(Student) 와 같다.
```
